```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("HR_Analytics.csv")
print("Dataset Loaded Successfully")
```

```python
print("\n================ FIRST 5 ROWS ================")
print(df.head())
print("\n================ LAST 5 ROWS ================")
print(df.tail())
print("\n================ SAMPLE RECORDS ================")
print(df.sample(5))
print("\n================ DATASET INFO ================")
print(df.info())
print("\n================ DATASET SHAPE ================")
print(df.shape)
print("\n================ COLUMN NAMES ================")
print(df.columns.tolist())
```

```python
print("\n================ MISSING VALUES ================")
print(df.isnull().sum())
print("\n================ DUPLICATE RECORDS ================")
print(df.duplicated().sum())
df = df.drop_duplicates()
print("\nDuplicates Removed Successfully")
print("\n================ DATA TYPES ================")
print(df.dtypes)
```

```python
print("\n================ NUMERICAL SUMMARY ================")
print(df.describe())
```

```python
print("\n================ KPI'S ================")
total_employees = df.shape[0]
average_salary = df["MonthlyIncome"].mean()
average_age = df["Age"].mean()
average_experience = df["TotalWorkingYears"].mean()
attrition_count = df["Attrition"].value_counts()
attrition_rate = (df["Attrition"]=="Yes").mean()*100
print("Total Employees:",total_employees)
print("Average Salary:",round(average_salary,2))
print("Average Age:",round(average_age,2))
print("Average Experience:",round(average_experience,2))
print("\nAttrition Count")
print(attrition_count)
print("\nAttrition Rate:",round(attrition_rate,2),"%")
```

```python
print("\n================ NUMPY ANALYSIS ================")
salary=np.array(df["MonthlyIncome"])
print("Mean :",np.mean(salary))
print("Median :",np.median(salary))
print("Minimum :",np.min(salary))
print("Maximum :",np.max(salary))
print("Standard Deviation :",np.std(salary))
print("Variance :",np.var(salary))
print("25 Percentile :",np.percentile(salary,25))
print("50 Percentile :",np.percentile(salary,50))
print("75 Percentile :",np.percentile(salary,75))
print("90 Percentile :",np.percentile(salary,90))
```

```python
print("\n================ DEPARTMENT VS ATTRITION ================")
department_attrition=pd.crosstab(df["Department"],df["Attrition"])
print(department_attrition)
sns.countplot(data=df,x="Department",hue="Attrition")
plt.title("Department vs Attrition")
plt.xticks(rotation=45)
plt.show()
```

```python
print("\n================ OVERTIME VS ATTRITION ================")
overtime=pd.crosstab(df["OverTime"],df["Attrition"])
print(overtime)
sns.countplot(data=df,x="OverTime",hue="Attrition")
plt.title("Overtime vs Attrition")
plt.show()
```

```python
print("\n================ AGE GROUP VS ATTRITION ================")
age=pd.crosstab(df["AgeGroup"],df["Attrition"])
print(age)
sns.countplot(data=df,x="AgeGroup",hue="Attrition")
plt.title("Age Group vs Attrition")
plt.show()
```

```python
print("\n================ SALARY VS ATTRITION ================")
salary=df.groupby("Attrition")["MonthlyIncome"].mean()
print(salary)
sns.boxplot(data=df,x="Attrition",y="MonthlyIncome")
plt.title("Salary vs Attrition")
plt.show()
```

```python
print("\n================ JOB ROLE VS ATTRITION ================")
jobrole=pd.crosstab(df["JobRole"],df["Attrition"])
print(jobrole)
sns.countplot(data=df,y="JobRole",hue="Attrition")
plt.title("Job Role vs Attrition")
plt.show()
```

```python
print("\n================ WORK LIFE BALANCE ================")
wlb=pd.crosstab(df["WorkLifeBalance"],df["Attrition"])
print(wlb)
sns.countplot(data=df,x="WorkLifeBalance",hue="Attrition")
plt.title("Work Life Balance vs Attrition")
plt.show()
```

```python
print("\n================ AVERAGE SALARY BY DEPARTMENT ================")
department_salary=df.groupby("Department")["MonthlyIncome"].mean().sort_values(ascending=False)
print(department_salary)
department_salary.plot(kind="bar")
plt.ylabel("Average Salary")
plt.title("Department Wise Average Salary")
plt.show()
```

```python
sns.scatterplot(data=df,x="TotalWorkingYears",y="MonthlyIncome")
plt.title("Experience vs Salary")
plt.show()
```

```python
print("\n================ CORRELATION ================")
numeric_df=df.select_dtypes(include="number")
correlation=numeric_df.corr()
print(correlation)
plt.figure(figsize=(15,10))
sns.heatmap(correlation,annot=True,cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```
